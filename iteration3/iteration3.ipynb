```python
import numpy as np

def get_center_and_size(pdb_file):
    coords = []
    with open(pdb_file) as f:
        for line in f:
            if line.startswith(("ATOM", "HETATM")):
                x = float(line[30:38])
                y = float(line[38:46])
                z = float(line[46:54])
                coords.append([x, y, z])
    coords = np.array(coords)
    center = coords.mean(axis=0)
    size = coords.max(axis=0) - coords.min(axis=0)
    return center, size

# Replace with your pdbqt file
center, size = get_center_and_size("2FOMiteration1.pdbqt")
print("Center:", center)
print("Size:", size)
```
